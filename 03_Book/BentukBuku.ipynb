```python
import cairo

# Dimensi kanvas potret tegak (1000 x 1400 px)
width, height = 1000, 1400
surface = cairo.ImageSurface(cairo.FORMAT_ARGB32, width, height)
ctx = cairo.Context(surface)

# Background kanvas
ctx.set_source_rgb(0.95, 0.95, 0.95)
ctx.paint()

# --- Shape Utama Sampul Buku (Biru Terang) ---
ctx.move_to(110, 60) # Titik Kiri-Atas

# Sisi Atas (Lengkung cembung tipis)
ctx.curve_to(350, 45, 650, 40, 890, 40)

# Sisi Kanan (Memanjang ke bawah mengikuti lekukan kertas)
ctx.curve_to(905, 450, 915, 850, 930, 1250)

# Sisi Bawah (Melengkung fleksibel)
ctx.curve_to(650, 1265, 350, 1255, 70, 1220)

# Sisi Kiri / Tulang Buku
ctx.curve_to(80, 850, 95, 450, 110, 60)

ctx.close_path()

# Isian Warna Biru Terang & Garis Outlinenya
ctx.set_source_rgb(0.0, 0.60, 0.82) # Biru terang
ctx.fill_preserve()

ctx.set_source_rgb(0.1, 0.1, 0.1)   # Outline tipis
ctx.set_line_width(3)
ctx.stroke()

surface.write_to_png("bentuk_buku_biru.png")
```
