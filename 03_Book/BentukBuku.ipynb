```python
import math
import cairo

width, height = 1000, 1400
surface = cairo.ImageSurface(cairo.FORMAT_ARGB32, width, height)
ctx = cairo.Context(surface)

# Background kanvas
ctx.set_source_rgb(0.95, 0.95, 0.95)
ctx.paint()

# 1. Sampul Luar Buku (Biru Terang)
ctx.move_to(110, 60)
ctx.curve_to(350, 45, 650, 40, 890, 40)
ctx.curve_to(905, 450, 915, 850, 930, 1250)
ctx.curve_to(650, 1265, 350, 1255, 70, 1220)
ctx.curve_to(80, 850, 95, 450, 110, 60)
ctx.close_path()

ctx.set_source_rgb(0.0, 0.60, 0.82)
ctx.fill_preserve()
ctx.set_source_rgb(0.1, 0.1, 0.1)
ctx.set_line_width(3)
ctx.stroke()

# 2. Teks "Edisi" & Angka "5" dengan Rotasi Perspektif
ctx.save()

# Titik acuan di pojok kanan atas buku
pos_x, pos_y = 815, 85
angle = math.radians(-2.5)  # Rotasi -2.5 derajat mengikuti sudut kemiringan atas buku

ctx.translate(pos_x, pos_y)
ctx.rotate(angle)

# --- Teks "Edisi" ---
ctx.select_font_face("Sans", cairo.FONT_SLANT_NORMAL, cairo.FONT_WEIGHT_BOLD)
ctx.set_font_size(30)
ctx.set_source_rgb(1.0, 1.0, 1.0)  # Putih

extents_edisi = ctx.text_extents("Edisi")
ctx.move_to(-extents_edisi.width / 2, 0)
ctx.show_text("Edisi")

# --- Angka "5" ---
ctx.set_font_size(115)
ctx.set_source_rgb(0.93, 0.40, 0.12)  # Warna oranye khas sampul

extents_5 = ctx.text_extents("5")
ctx.move_to(-extents_5.width / 2, 95)  # Diletakkan tepat di bawah "Edisi"
ctx.show_text("5")

ctx.restore()

surface.write_to_png("bentuk_buku_edisi5.png")
```
